```python
# TASK 2: DATA PREPARATION & PREPROCESSING

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Remove ID and Separate Target
X = df.drop(
    columns=["Patient_ID", "Readmitted_30_Days"]
)

y = df["Readmitted_30_Days"]

# Identify Feature Types
numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))

# Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# Preprocessing Pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            StandardScaler(),
            numeric_features
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ]
)

# Fit ONLY on Training Data
preprocessor.fit(X_train)

# Transform Train and Test
X_train_processed = preprocessor.transform(X_train)
X_test_processed = preprocessor.transform(X_test)

# Feature Names
feature_names = preprocessor.get_feature_names_out()

# Final Results
print("\nOriginal feature count:", X.shape[1])
print("Processed feature count:", len(feature_names))

print(
    "\nProcessed training shape:",
    X_train_processed.shape
)

print(
    "Processed testing shape:",
    X_test_processed.shape
)

print("\nPreprocessing completed successfully!")
```

```python
# TASK 3: TRAIN THREE CLASSIFICATION MODELS

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

# 1. LOGISTIC REGRESSION
logistic_model = LogisticRegression(
    C=1.0,
    max_iter=1000,
    solver="liblinear",
    random_state=42
)

logistic_model.fit(
    X_train_processed,
    y_train
)

# 2. KNN
knn_model = KNeighborsClassifier(
    n_neighbors=5
)

knn_model.fit(
    X_train_processed,
    y_train
)

# 3. DECISION TRE
decision_tree_model = DecisionTreeClassifier(
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42
)

decision_tree_model.fit(
    X_train_processed,
    y_train
)

print("All three models trained successfully!")

# 4. TRAINING PREDICTIONS
lr_train_pred = logistic_model.predict(
    X_train_processed
)

knn_train_pred = knn_model.predict(
    X_train_processed
)

dt_train_pred = decision_tree_model.predict(
    X_train_processed
)

# 5. TEST PREDICTIONS
lr_test_pred = logistic_model.predict(
    X_test_processed
)

knn_test_pred = knn_model.predict(
    X_test_processed
)

dt_test_pred = decision_tree_model.predict(
    X_test_processed
)

# 6. TEST PROBABILITIES
lr_test_proba = logistic_model.predict_proba(
    X_test_processed
)[:, 1]

knn_test_proba = knn_model.predict_proba(
    X_test_processed
)[:, 1]

dt_test_proba = decision_tree_model.predict_proba(
    X_test_processed
)[:, 1]

# 7. Verify
print("\nPrediction sizes:")

print(
    "Logistic Regression:",
    len(lr_test_pred)
)

print(
    "KNN:",
    len(knn_test_pred)
)

print(
    "Decision Tree:",
    len(dt_test_pred)
)
```

```python
# TASK 4: MODEL EVALUATION

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

# 2. VERIFY MODELS AND DATA
print("=" * 70)
print("DATA VERIFICATION")
print("=" * 70)

print("\nTraining data shape:")
print(X_train_processed.shape)

print("\nTesting data shape:")
print(X_test_processed.shape)

print("\nTraining target shape:")
print(y_train.shape)

print("\nTesting target shape:")
print(y_test.shape)

print("\nModels:")
print("1. Logistic Regression")
print("2. KNN")
print("3. Decision Tree")

# 3. GENERATE TRAINING PREDICTIONS
# Logistic Regression
lr_train_pred = logistic_model.predict(
    X_train_processed
)

# KNN
knn_train_pred = knn_model.predict(
    X_train_processed
)

# Decision Tree
dt_train_pred = decision_tree_model.predict(
    X_train_processed
)

# 4. GENERATE TEST PREDICTIONS
# Logistic Regression
lr_test_pred = logistic_model.predict(
    X_test_processed
)

# KNN
knn_test_pred = knn_model.predict(
    X_test_processed
)

# Decision Tree
dt_test_pred = decision_tree_model.predict(
    X_test_processed
)

# 5. GENERATE TEST PROBABILITIES
lr_test_proba = logistic_model.predict_proba(
    X_test_processed
)[:, 1]

knn_test_proba = knn_model.predict_proba(
    X_test_processed
)[:, 1]

dt_test_proba = decision_tree_model.predict_proba(
    X_test_processed
)[:, 1]

# 6. CREATE EVALUATION FUNCTION
def evaluate_model(
    model_name,
    y_train,
    train_predictions,
    y_test,
    test_predictions,
    test_probabilities
):

    # Training Accuracy
    train_accuracy = accuracy_score(
        y_train,
        train_predictions
    )

    # Test Accuracy
    test_accuracy = accuracy_score(
        y_test,
        test_predictions
    )

    # Precision
    precision = precision_score(
        y_test,
        test_predictions,
        zero_division=0
    )

    # Recall
    recall = recall_score(
        y_test,
        test_predictions,
        zero_division=0
    )

    # F1 Score
    f1 = f1_score(
        y_test,
        test_predictions,
        zero_division=0
    )

    # ROC-AUC
    roc_auc = roc_auc_score(
        y_test,
        test_probabilities
    )

    # Confusion Matrix
    cm = confusion_matrix(
        y_test,
        test_predictions
    )

    # Generalization Gap
    generalization_gap = (
        train_accuracy - test_accuracy
    )

    # Store result
    return {
        "Model": model_name,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "ROC-AUC": roc_auc,
        "Generalization Gap": generalization_gap,
        "Confusion Matrix": cm
    }

# 7. EVALUATE LOGISTIC REGRESSION
lr_results = evaluate_model(
    model_name="Logistic Regression",

    y_train=y_train,
    train_predictions=lr_train_pred,

    y_test=y_test,
    test_predictions=lr_test_pred,

    test_probabilities=lr_test_proba
)

# 8. EVALUATE KNN
knn_results = evaluate_model(
    model_name="KNN",

    y_train=y_train,
    train_predictions=knn_train_pred,

    y_test=y_test,
    test_predictions=knn_test_pred,

    test_probabilities=knn_test_proba
)

# 9. EVALUATE DECISION TREE
dt_results = evaluate_model(
    model_name="Decision Tree",

    y_train=y_train,
    train_predictions=dt_train_pred,

    y_test=y_test,
    test_predictions=dt_test_pred,

    test_probabilities=dt_test_proba
)

# 10. STORE ALL RESULTS
all_results = [
    lr_results,
    knn_results,
    dt_results
]

# 11. PRINT INDIVIDUAL MODEL RESULTS
for result in all_results:

    print("\n")
    print("=" * 70)
    print(result["Model"])
    print("=" * 70)

    print(
        f"Train Accuracy       : "
        f"{result['Train Accuracy']:.4f}"
    )

    print(
        f"Test Accuracy        : "
        f"{result['Test Accuracy']:.4f}"
    )

    print(
        f"Precision            : "
        f"{result['Precision']:.4f}"
    )

    print(
        f"Recall               : "
        f"{result['Recall']:.4f}"
    )

    print(
        f"F1-Score             : "
        f"{result['F1-Score']:.4f}"
    )

    print(
        f"ROC-AUC              : "
        f"{result['ROC-AUC']:.4f}"
    )

    print(
        f"Generalization Gap   : "
        f"{result['Generalization Gap']:.4f}"
    )

    print("\nConfusion Matrix:")

    print(
        result["Confusion Matrix"]
    )

# 12. CREATE MODEL COMPARISON DATAFRAME
comparison_df = pd.DataFrame({

    "Model": [
        "Logistic Regression",
        "KNN",
        "Decision Tree"
    ],

    "Train Accuracy": [
        lr_results["Train Accuracy"],
        knn_results["Train Accuracy"],
        dt_results["Train Accuracy"]
    ],

    "Test Accuracy": [
        lr_results["Test Accuracy"],
        knn_results["Test Accuracy"],
        dt_results["Test Accuracy"]
    ],

    "Precision": [
        lr_results["Precision"],
        knn_results["Precision"],
        dt_results["Precision"]
    ],

    "Recall": [
        lr_results["Recall"],
        knn_results["Recall"],
        dt_results["Recall"]
    ],

    "F1-Score": [
        lr_results["F1-Score"],
        knn_results["F1-Score"],
        dt_results["F1-Score"]
    ],

    "ROC-AUC": [
        lr_results["ROC-AUC"],
        knn_results["ROC-AUC"],
        dt_results["ROC-AUC"]
    ],

    "Generalization Gap": [
        lr_results["Generalization Gap"],
        knn_results["Generalization Gap"],
        dt_results["Generalization Gap"]
    ]
})

# 13. DISPLAY COMPARISON TABLE
print("\n")
print("=" * 100)
print("MODEL COMPARISON TABLE")
print("=" * 100)

display(
    comparison_df.style.format({
        "Train Accuracy": "{:.4f}",
        "Test Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "ROC-AUC": "{:.4f}",
        "Generalization Gap": "{:.4f}"
    })
)

# 14. CONFUSION MATRIX - LOGISTIC REGRESSION
lr_cm = lr_results["Confusion Matrix"]

plt.figure(figsize=(7, 5))

sns.heatmap(
    lr_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Not Readmitted",
        "Readmitted"
    ],
    yticklabels=[
        "Not Readmitted",
        "Readmitted"
    ]
)

plt.title(
    "Logistic Regression - Confusion Matrix"
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()
plt.show()

# 15. CONFUSION MATRIX - KNN
knn_cm = knn_results["Confusion Matrix"]

plt.figure(figsize=(7, 5))

sns.heatmap(
    knn_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Not Readmitted",
        "Readmitted"
    ],
    yticklabels=[
        "Not Readmitted",
        "Readmitted"
    ]
)

plt.title(
    "KNN - Confusion Matrix"
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()
plt.show()

# 16. CONFUSION MATRIX - DECISION TREE
dt_cm = dt_results["Confusion Matrix"]

plt.figure(figsize=(7, 5))

sns.heatmap(
    dt_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Not Readmitted",
        "Readmitted"
    ],
    yticklabels=[
        "Not Readmitted",
        "Readmitted"
    ]
)

plt.title(
    "Decision Tree - Confusion Matrix"
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()
plt.show()

# 17. PRINT TP, TN, FP, FN\
def print_confusion_matrix_details(
    model_name,
    cm
):

    tn, fp, fn, tp = cm.ravel()

    print("\n")
    print("=" * 60)
    print(model_name)
    print("=" * 60)

    print(f"True Negative  (TN): {tn}")
    print(f"False Positive (FP): {fp}")
    print(f"False Negative (FN): {fn}")
    print(f"True Positive  (TP): {tp}")


print_confusion_matrix_details(
    "Logistic Regression",
    lr_cm
)

print_confusion_matrix_details(
    "KNN",
    knn_cm
)

print_confusion_matrix_details(
    "Decision Tree",
    dt_cm
)

# 18. CLASSIFICATION REPORTS
print("\n")
print("=" * 70)
print("LOGISTIC REGRESSION CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        lr_test_pred,
        target_names=[
            "Not Readmitted",
            "Readmitted"
        ],
        zero_division=0
    )
)


print("\n")
print("=" * 70)
print("KNN CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        knn_test_pred,
        target_names=[
            "Not Readmitted",
            "Readmitted"
        ],
        zero_division=0
    )
)


print("\n")
print("=" * 70)
print("DECISION TREE CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        dt_test_pred,
        target_names=[
            "Not Readmitted",
            "Readmitted"
        ],
        zero_division=0
    )
)

# 19. METRIC COMPARISON CHART
metrics = [
    "Test Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "ROC-AUC"
]

plot_df = comparison_df[
    ["Model"] + metrics
]

plot_df_melted = plot_df.melt(
    id_vars="Model",
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=plot_df_melted,
    x="Metric",
    y="Score",
    hue="Model"
)

plt.title(
    "Model Performance Comparison"
)

plt.xlabel("Evaluation Metric")
plt.ylabel("Score")

plt.ylim(0, 1)

plt.legend(
    title="Model"
)

plt.tight_layout()
plt.show()

# 20. TRAINING VS TEST ACCURACY
train_test_df = comparison_df[
    [
        "Model",
        "Train Accuracy",
        "Test Accuracy"
    ]
]

train_test_melted = train_test_df.melt(
    id_vars="Model",
    var_name="Dataset",
    value_name="Accuracy"
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=train_test_melted,
    x="Model",
    y="Accuracy",
    hue="Dataset"
)

plt.title(
    "Training vs Test Accuracy"
)

plt.xlabel("Model")
plt.ylabel("Accuracy")

plt.ylim(0, 1)

plt.legend(
    title="Dataset"
)

plt.tight_layout()
plt.show()

# 21. GENERALIZATION GAP CHART
plt.figure(figsize=(9, 5))

sns.barplot(
    data=comparison_df,
    x="Model",
    y="Generalization Gap"
)

plt.axhline(
    y=0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "Train-Test Generalization Gap"
)

plt.xlabel("Model")
plt.ylabel(
    "Train Accuracy - Test Accuracy"
)

plt.tight_layout()
plt.show()

# 22. ROC CURVES
# Logistic Regression ROC
lr_fpr, lr_tpr, _ = roc_curve(
    y_test,
    lr_test_proba
)

# KNN ROC
knn_fpr, knn_tpr, _ = roc_curve(
    y_test,
    knn_test_proba
)

# Decision Tree ROC
dt_fpr, dt_tpr, _ = roc_curve(
    y_test,
    dt_test_proba
)


# Plot
plt.figure(figsize=(9, 7))

plt.plot(
    lr_fpr,
    lr_tpr,
    label=f"Logistic Regression "
          f"(AUC = {lr_results['ROC-AUC']:.4f})"
)

plt.plot(
    knn_fpr,
    knn_tpr,
    label=f"KNN "
          f"(AUC = {knn_results['ROC-AUC']:.4f})"
)

plt.plot(
    dt_fpr,
    dt_tpr,
    label=f"Decision Tree "
          f"(AUC = {dt_results['ROC-AUC']:.4f})"
)

# Random classifier reference
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.title(
    "ROC Curves - Model Comparison"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# 23. FINAL FORMATTED TABLE
final_comparison = comparison_df.copy()

percentage_columns = [
    "Train Accuracy",
    "Test Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "ROC-AUC",
    "Generalization Gap"
]

for column in percentage_columns:

    final_comparison[column] = (
        final_comparison[column] * 100
    ).round(2)


print("\n")
print("=" * 100)
print("FINAL MODEL COMPARISON (%)")
print("=" * 100)

display(final_comparison)

# 24. AUTOMATIC FIT DIAGNOSIS
print("\n")
print("=" * 80)
print("INITIAL FIT DIAGNOSIS")
print("=" * 80)

for _, row in comparison_df.iterrows():

    model = row["Model"]

    train_acc = row["Train Accuracy"]
    test_acc = row["Test Accuracy"]

    gap = row["Generalization Gap"]

    print(f"\n{model}")

    print(
        f"Train Accuracy: {train_acc:.4f}"
    )

    print(
        f"Test Accuracy : {test_acc:.4f}"
    )

    print(
        f"Gap           : {gap:.4f}"
    )

    # Simple diagnostic rule
    if gap >= 0.05:

        print(
            "Diagnosis: Noticeable train-test gap; "
            "possible overfitting/generalization issue."
        )

    elif train_acc < 0.75 and test_acc < 0.75:

        print(
            "Diagnosis: Low training and test performance; "
            "possible underfitting."
        )

    else:

        print(
            "Diagnosis: No strong train-test accuracy gap "
            "based on this baseline evaluation."
        )

# 25. SAVE COMPARISON TABLE
comparison_df.to_csv(
    "task4_model_comparison.csv",
    index=False
)
```

```python
# TASK 5: HYPERPARAMETER TUNING AND MODEL SELECTION

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

import pandas as pd
import matplotlib.pyplot as plt

# 1. Create 5-Fold Stratified Cross Validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# 2. LOGISTIC REGRESSION HYPERPARAMETER TUNING
logistic_pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LogisticRegression(
        max_iter=2000,
        solver="liblinear",
        random_state=42
    ))
])

logistic_params = {
    "model__C": [0.01, 0.1, 1, 10, 100]
}

logistic_grid = GridSearchCV(
    logistic_pipeline,
    logistic_params,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1,
    return_train_score=True
)

logistic_grid.fit(X_train, y_train)

print("LOGISTIC REGRESSION")
print("Best Parameters:", logistic_grid.best_params_)
print("Best CV ROC-AUC:", logistic_grid.best_score_)

# 3. KNN HYPERPARAMETER TUNING
knn_pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", KNeighborsClassifier())
])

knn_params = {
    "model__n_neighbors": [1, 3, 5, 7, 9, 11, 15, 21]
}

knn_grid = GridSearchCV(
    knn_pipeline,
    knn_params,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1,
    return_train_score=True
)

knn_grid.fit(X_train, y_train)

print("\nKNN")
print("Best Parameters:", knn_grid.best_params_)
print("Best CV ROC-AUC:", knn_grid.best_score_)

# 4. DECISION TREE HYPERPARAMETER TUNING
tree_pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", DecisionTreeClassifier(
        random_state=42
    ))
])

tree_params = {
    "model__max_depth": [3, 5, 7, 10, 15, None],
    "model__min_samples_leaf": [1, 5, 10]
}

tree_grid = GridSearchCV(
    tree_pipeline,
    tree_params,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1,
    return_train_score=True
)

tree_grid.fit(X_train, y_train)

print("\nDECISION TREE")
print("Best Parameters:", tree_grid.best_params_)
print("Best CV ROC-AUC:", tree_grid.best_score_)

# 5. STORE THE BEST MODELS
final_logistic_model = logistic_grid.best_estimator_
final_knn_model = knn_grid.best_estimator_
final_tree_model = tree_grid.best_estimator_

# 6. DISPLAY ALL LOGISTIC REGRESSION RESULTS
logistic_results = pd.DataFrame(logistic_grid.cv_results_)

print("\nLogistic Regression Tuning Results:")
print(
    logistic_results[
        [
            "param_model__C",
            "mean_train_score",
            "mean_test_score"
        ]
    ]
)

# 7. DISPLAY ALL KNN RESULTS
knn_results = pd.DataFrame(knn_grid.cv_results_)

print("\nKNN Tuning Results:")
print(
    knn_results[
        [
            "param_model__n_neighbors",
            "mean_train_score",
            "mean_test_score"
        ]
    ]
)

# 8. DISPLAY ALL DECISION TREE RESULTS
tree_results = pd.DataFrame(tree_grid.cv_results_)

print("\nDecision Tree Tuning Results:")
print(
    tree_results[
        [
            "param_model__max_depth",
            "param_model__min_samples_leaf",
            "mean_train_score",
            "mean_test_score"
        ]
    ]
)

# 9. COMPARE BEST PARAMETERS
print("FINAL SELECTED PARAMETERS")

print("\nLogistic Regression:")
print(logistic_grid.best_params_)

print("\nKNN:")
print(knn_grid.best_params_)

print("\nDecision Tree:")
print(tree_grid.best_params_)

# 10. CREATE SUMMARY TABLE
tuning_summary = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "KNN",
        "Decision Tree"
    ],
    "Best Parameters": [
        str(logistic_grid.best_params_),
        str(knn_grid.best_params_),
        str(tree_grid.best_params_)
    ],
    "Best CV ROC-AUC": [
        logistic_grid.best_score_,
        knn_grid.best_score_,
        tree_grid.best_score_
    ]
})

print("\nHyperparameter Tuning Summary:")
print(tuning_summary)

# 11. PLOT LOGISTIC REGRESSION TUNING
plt.figure(figsize=(8, 5))

plt.plot(
    logistic_results["param_model__C"].astype(float),
    logistic_results["mean_test_score"],
    marker="o"
)

plt.xscale("log")
plt.xlabel("C")
plt.ylabel("Mean CV ROC-AUC")
plt.title("Logistic Regression Hyperparameter Tuning")
plt.show()

# 12. PLOT KNN TUNING
plt.figure(figsize=(8, 5))

plt.plot(
    knn_results["param_model__n_neighbors"].astype(int),
    knn_results["mean_test_score"],
    marker="o"
)

plt.xlabel("Number of Neighbors (K)")
plt.ylabel("Mean CV ROC-AUC")
plt.title("KNN Hyperparameter Tuning")
plt.show()

# 13. FINAL MESSAGE
print("\nHyperparameter tuning completed successfully.")
print("The best models have been stored as:")

print("final_logistic_model")
print("final_knn_model")
print("final_tree_model")
```

```python
# TASK 6: FINAL EVALUATION OF TUNED MODELS

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

import pandas as pd
import matplotlib.pyplot as plt

# 1. Make predictions using the tuned models
# Logistic Regression
logistic_train_pred = final_logistic_model.predict(X_train)
logistic_test_pred = final_logistic_model.predict(X_test)

logistic_train_prob = final_logistic_model.predict_proba(X_train)[:, 1]
logistic_test_prob = final_logistic_model.predict_proba(X_test)[:, 1]


# KNN
knn_train_pred = final_knn_model.predict(X_train)
knn_test_pred = final_knn_model.predict(X_test)

knn_train_prob = final_knn_model.predict_proba(X_train)[:, 1]
knn_test_prob = final_knn_model.predict_proba(X_test)[:, 1]


# Decision Tree
tree_train_pred = final_tree_model.predict(X_train)
tree_test_pred = final_tree_model.predict(X_test)

tree_train_prob = final_tree_model.predict_proba(X_train)[:, 1]
tree_test_prob = final_tree_model.predict_proba(X_test)[:, 1]

# 2. Calculate metrics for each model
# Logistic Regression
logistic_train_accuracy = accuracy_score(y_train, logistic_train_pred)
logistic_test_accuracy = accuracy_score(y_test, logistic_test_pred)

logistic_precision = precision_score(y_test, logistic_test_pred)
logistic_recall = recall_score(y_test, logistic_test_pred)
logistic_f1 = f1_score(y_test, logistic_test_pred)
logistic_auc = roc_auc_score(y_test, logistic_test_prob)


# KNN
knn_train_accuracy = accuracy_score(y_train, knn_train_pred)
knn_test_accuracy = accuracy_score(y_test, knn_test_pred)

knn_precision = precision_score(y_test, knn_test_pred)
knn_recall = recall_score(y_test, knn_test_pred)
knn_f1 = f1_score(y_test, knn_test_pred)
knn_auc = roc_auc_score(y_test, knn_test_prob)


# Decision Tree
tree_train_accuracy = accuracy_score(y_train, tree_train_pred)
tree_test_accuracy = accuracy_score(y_test, tree_test_pred)

tree_precision = precision_score(y_test, tree_test_pred)
tree_recall = recall_score(y_test, tree_test_pred)
tree_f1 = f1_score(y_test, tree_test_pred)
tree_auc = roc_auc_score(y_test, tree_test_prob)

# 3. Create comparison table
comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "KNN",
        "Decision Tree"
    ],

    "Train Accuracy": [
        logistic_train_accuracy,
        knn_train_accuracy,
        tree_train_accuracy
    ],

    "Test Accuracy": [
        logistic_test_accuracy,
        knn_test_accuracy,
        tree_test_accuracy
    ],

    "Precision": [
        logistic_precision,
        knn_precision,
        tree_precision
    ],

    "Recall": [
        logistic_recall,
        knn_recall,
        tree_recall
    ],

    "F1 Score": [
        logistic_f1,
        knn_f1,
        tree_f1
    ],

    "ROC-AUC": [
        logistic_auc,
        knn_auc,
        tree_auc
    ]
})

# 4. Calculate train-test accuracy gap
comparison["Accuracy Gap"] = (
    comparison["Train Accuracy"] -
    comparison["Test Accuracy"])

# 5. Fit diagnosis
def diagnose_model(gap):

    if gap >= 0.05:
        return "Possible Overfitting"

    elif gap >= 0.02:
        return "Moderate Generalization Gap"

    else:
        return "Good Generalization"


comparison["Fit Diagnosis"] = comparison["Accuracy Gap"].apply(
    diagnose_model
)

# 6. Display comparison tabl
print("\nFINAL MODEL COMPARISON")

print(comparison.round(4).to_string(index=False))

# 7. Display percentages
print("\nFINAL MODEL COMPARISON IN PERCENTAGE")
percentage_comparison = comparison.copy()

percentage_columns = [
    "Train Accuracy",
    "Test Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"
]

for column in percentage_columns:
    percentage_comparison[column] = (
        percentage_comparison[column] * 100
    ).round(2)

percentage_comparison["Accuracy Gap"] = (
    percentage_comparison["Accuracy Gap"] * 100
).round(2)

print(
    percentage_comparison.to_string(index=False)
)

# 8. Logistic Regression Classification Report
print("\nLOGISTIC REGRESSION CLASSIFICATION REPORT")

print(classification_report(
        y_test,
        logistic_test_pred,
        target_names=[
            "Not Readmitted",
            "Readmitted"]))

# 9. KNN Classification Report
print("\nKNN CLASSIFICATION REPORT")

print(classification_report(
        y_test,
        knn_test_pred,
        target_names=[
            "Not Readmitted",
            "Readmitted"]))

# 10. Decision Tree Classification Report
print("\nDECISION TREE CLASSIFICATION REPORT")
print(classification_report(
        y_test,
        tree_test_pred,
        target_names=[
            "Not Readmitted",
            "Readmitted"]))

# 11. Confusion Matrix - Logistic Regression
logistic_cm = confusion_matrix(
    y_test,
    logistic_test_pred
)

print("\nLogistic Regression Confusion Matrix:")
print(logistic_cm)

print("TN =", logistic_cm[0, 0])
print("FP =", logistic_cm[0, 1])
print("FN =", logistic_cm[1, 0])
print("TP =", logistic_cm[1, 1])

# 12. Confusion Matrix - KNN
knn_cm = confusion_matrix(
    y_test,
    knn_test_pred
)

print("\nKNN Confusion Matrix:")
print(knn_cm)

print("TN =", knn_cm[0, 0])
print("FP =", knn_cm[0, 1])
print("FN =", knn_cm[1, 0])
print("TP =", knn_cm[1, 1])

# 13. Confusion Matrix - Decision Tree
tree_cm = confusion_matrix(
    y_test,
    tree_test_pred
)

print("\nDecision Tree Confusion Matrix:")
print(tree_cm)

print("TN =", tree_cm[0, 0])
print("FP =", tree_cm[0, 1])
print("FN =", tree_cm[1, 0])
print("TP =", tree_cm[1, 1])

# 14. Plot confusion matrices
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].imshow(logistic_cm)
axes[0].set_title("Logistic Regression")
axes[0].set_xlabel("Predicted")
axes[0].set_ylabel("Actual")

for i in range(2):
    for j in range(2):
        axes[0].text(
            j,
            i,
            logistic_cm[i, j],
            ha="center",
            va="center"
        )

axes[0].set_xticks([0, 1])
axes[0].set_yticks([0, 1])


axes[1].imshow(knn_cm)
axes[1].set_title("KNN")
axes[1].set_xlabel("Predicted")
axes[1].set_ylabel("Actual")

for i in range(2):
    for j in range(2):
        axes[1].text(
            j,
            i,
            knn_cm[i, j],
            ha="center",
            va="center"
        )

axes[1].set_xticks([0, 1])
axes[1].set_yticks([0, 1])


axes[2].imshow(tree_cm)
axes[2].set_title("Decision Tree")
axes[2].set_xlabel("Predicted")
axes[2].set_ylabel("Actual")

for i in range(2):
    for j in range(2):
        axes[2].text(
            j,
            i,
            tree_cm[i, j],
            ha="center",
            va="center"
        )

axes[2].set_xticks([0, 1])
axes[2].set_yticks([0, 1])

plt.tight_layout()
plt.show()

# 15. Metric Comparison Graph
models = [
    "Logistic Regression",
    "KNN",
    "Decision Tree"
]

x = range(len(models))

plt.figure(figsize=(10, 6))

plt.plot(
    x,
    comparison["Test Accuracy"],
    marker="o",
    label="Accuracy"
)

plt.plot(
    x,
    comparison["Precision"],
    marker="o",
    label="Precision"
)

plt.plot(
    x,
    comparison["Recall"],
    marker="o",
    label="Recall"
)

plt.plot(
    x,
    comparison["F1 Score"],
    marker="o",
    label="F1 Score"
)

plt.plot(
    x,
    comparison["ROC-AUC"],
    marker="o",
    label="ROC-AUC"
)

plt.xticks(x, models)
plt.ylabel("Score")
plt.title("Comparison of Model Performance")
plt.legend()
plt.grid()
plt.show()

# 16. ROC Curves
logistic_fpr, logistic_tpr, _ = roc_curve(
    y_test,
    logistic_test_prob
)

knn_fpr, knn_tpr, _ = roc_curve(
    y_test,
    knn_test_prob
)

tree_fpr, tree_tpr, _ = roc_curve(
    y_test,
    tree_test_prob
)


plt.figure(figsize=(8, 6))

plt.plot(
    logistic_fpr,
    logistic_tpr,
    label="Logistic Regression"
)

plt.plot(
    knn_fpr,
    knn_tpr,
    label="KNN"
)

plt.plot(
    tree_fpr,
    tree_tpr,
    label="Decision Tree"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()
plt.grid()
plt.show()

# 17. Train vs Test Accuracy
plt.figure(figsize=(8, 5))

plt.plot(
    models,
    comparison["Train Accuracy"],
    marker="o",
    label="Train Accuracy"
)

plt.plot(
    models,
    comparison["Test Accuracy"],
    marker="o",
    label="Test Accuracy"
)

plt.ylabel("Accuracy")
plt.title("Train vs Test Accuracy")
plt.legend()
plt.grid()
plt.show()

# 18. Print accuracy gaps
print("\nTRAIN VS TEST ACCURACY GAP")
for i in range(len(comparison)):

    model = comparison.loc[i, "Model"]
    train_accuracy = comparison.loc[i, "Train Accuracy"]
    test_accuracy = comparison.loc[i, "Test Accuracy"]
    gap = comparison.loc[i, "Accuracy Gap"]

    print("\nModel:", model)
    print("Train Accuracy:", round(train_accuracy * 100, 2), "%")
    print("Test Accuracy:", round(test_accuracy * 100, 2), "%")
    print("Accuracy Gap:", round(gap * 100, 2), "%")
    print("Diagnosis:", comparison.loc[i, "Fit Diagnosis"])

# 19. Final tuned model parameters
print("\nFINAL TUNED MODEL PARAMETERS")

print("\nLogistic Regression:")
print(logistic_grid.best_params_)

print("\nKNN:")
print(knn_grid.best_params_)

print("\nDecision Tree:")
print(tree_grid.best_params_)
```
